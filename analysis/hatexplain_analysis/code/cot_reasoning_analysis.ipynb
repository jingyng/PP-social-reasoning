```python
import pandas as pd

df = pd.read_csv("extracted_cot_reasoning.csv")
```

```python
# Allow pandas to display all columns
pd.set_option('display.max_columns', None)

# Allow pandas to display more rows
pd.set_option('display.max_rows', 100)

# **Most Important for you**: Increase the column width to read the CoT
pd.set_option('display.max_colwidth', 200)

# Loop through each row in the DataFrame
for index, row in df.iterrows():
    if "s0428.jsonl" not in row["run_name"]:
        continue
    # For the current row, loop through each column and its value
    for col_name, value in row.items():
        # Print the column name and its value, separated by a line break
        print(f"[{col_name}]:")
        print(value)
        print() # Add a small space for clarity
    print(f"--- Row {index} ---") # Print the row index as a header
    print("=" * 50) # Print a clear separator after each row is finished
```
